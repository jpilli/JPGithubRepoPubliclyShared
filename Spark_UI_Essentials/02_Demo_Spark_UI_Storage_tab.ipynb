```python
# Create a sample DataFrame
df = spark.range(0, 10000000).selectExpr(
    "id",
    "id * 2 as doubled",
    "concat('user_', id) as username"
)

# Cache (default: MEMORY_AND_DISK)
df_cached = df.cache()

# Trigger an action to materialize the cache
count = df_cached.count()
```

```python
# To clear cache when done
df_cached.unpersist()
```
