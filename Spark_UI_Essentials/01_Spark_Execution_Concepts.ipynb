#Spark Execution Concepts - Demo

##Query 1: Narrow transformation

```python
#Query 1: Narrow transformation
df_customer = (
    spark.read.table("samples.tpch.customer")
    .select("c_custkey", "c_name", "c_acctbal", "c_mktsegment")
    .filter("c_nationkey == 21")
)

display(df_customer)
```

##Query 2a: Wide transformation (AQE disabled)

```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
```

```python
#Query 2a: Wide transformation (AQE disabled)
from pyspark.sql import functions as F

df_customer = (
    spark.read.table("samples.tpch.customer")
    .groupBy("c_nationkey")
    .agg(F.sum("c_acctbal").alias("total_acctbal"))
)

display(df_customer)
```

##Query 2b: Wide transformation (AQE ENABLED (default))

AQE: Adaptive Query Execution reoptimizes and adjusts query plans based on runtime statistics (rather than static statistics) 

```python
spark.conf.set("spark.sql.adaptive.enabled", "true")
```

```python
#Query 2b: Wide transformation (AQE enabled)
from pyspark.sql import functions as F

df_customer = (
    spark.read.table("samples.tpch.customer")
    .groupBy("c_nationkey")
    .agg(F.sum("c_acctbal").alias("total_acctbal"))
)

display(df_customer)
```
